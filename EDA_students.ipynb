```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('bi.csv', encoding='windows-1251')
print(df)
```

```python
print("Размер датасета:", df.shape)
print(" Заголовки столбцов:")
print(df.columns.tolist())
```

```python
print("Первые 4 строки:")
df.head(4)
```

```python
print("Последние 10 строк:")
df.tail(10)
```

```python
print("Сводная информация:")
df.info()
```

```python
print("АНАЛИЗ ПРОПУЩЕННЫХ ЗНАЧЕНИЙ:")
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
```

```python
numerical_cols = df.select_dtypes(include=[np.number]).columns

print("\nАНАЛИЗ ЧИСЛОВЫХ ПЕРЕМЕННЫХ:")
for col in numerical_cols:
    print(f"\n--- {col} ---")
    print(f"Среднее: {df[col].mean():.2f}")
    print(f"Медиана: {df[col].median():.2f}")
    print(f"Стандартное отклонение: {df[col].std():.2f}")
    print(f"Асимметрия: {df[col].skew():.2f}")
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns

print("\nАНАЛИЗ КАТЕГОРИАЛЬНЫХ ПЕРЕМЕННЫХ:")
for col in categorical_cols:
    print(f"\n--- {col} ---")
    value_counts = df[col].value_counts()
    print(value_counts.head(10))
```

```python
if len(numerical_cols) > 1:
    print("\nМАТРИЦА КОРРЕЛЯЦИЙ:")
    correlation_matrix = df[numerical_cols].corr()
    print(correlation_matrix)

    plt.figure(figsize=(10, 8))
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
    plt.title("Матрица корреляций")
    plt.show()
```

```python
print("\nАНАЛИЗ ВЫБРОСОВ:")
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} выбросов ({len(outliers)/len(df)*100:.1f}%)")
```

```python
print("\nВИЗУАЛИЗАЦИЯ РАСПРЕДЕЛЕНИЯ ВОЗРАСТОВ:")
plt.figure(figsize=(10, 5))
sns.histplot(df['Age'], bins=15, kde=True, color='salmon')
plt.title('Распределение возрастов студентов', fontsize=14)
plt.xlabel('Возраст')
plt.ylabel('Количество студентов')
plt.axvline(df['Age'].mean(), color='red', linestyle='--', label=f'Среднее: {df["Age"].mean():.1f}')
plt.legend()
plt.show()
```
